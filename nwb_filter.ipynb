```python
from pynwb import NWBHDF5IO

path = "data/AO026_20181122_180943.nwb"

io = NWBHDF5IO(path, "r", load_namespacesgit=True)
nwb = io.read()

print("\n=== ACQUISITION ===")
print(list(nwb.acquisition.keys()))

print("\n=== PROCESSING ===")
print(list(nwb.processing.keys()))

print("\n=== TRIALS ===")
print("has trials:", nwb.trials is not None)
if nwb.trials is not None:
    print("trial columns:", list(nwb.trials.colnames))

print("\n=== INTERVALS ===")
print(list(nwb.intervals.keys()))

print("\n=== UNITS ===")
print("has units:", nwb.units is not None)
if nwb.units is not None:
    print("unit columns:", list(nwb.units.colnames))
    print("n_units:", len(nwb.units.id[:]))
```
